```python
import mortapy as mp
import math # Diperlukan jika Anda ingin melakukan verifikasi manual yang melibatkan math.exp
# from IPython.display import display, Markdown # Tidak diperlukan jika hanya print dan .show()

print("Library mortapy berhasil diimpor.")
print("Contoh ini akan mendemonstrasikan perhitungan berbasis asumsi distribusi murni,")
print("termasuk Gompertz dan Makeham.")
```

```python
# Parameter Umum yang akan digunakan untuk berbagai contoh
usia_awal = 40
periode_hidup_bulat = 5  # tahun
periode_hidup_non_bulat = 5.75 # tahun
suku_bunga = 0.04 # 4%

# Parameter untuk asumsi Gompertz
B_gomp = 0.00006
c_gomp = 1.085

# Parameter untuk asumsi Makeham
A_make = 0.00015
B_make = 0.00004
c_make = 1.095

print("Parameter pengujian telah diatur.")
```

```python
print("\n" + "="*60)
print("PENGUJIAN PROBABILITAS HIDUP DENGAN BERBAGAI ASUMSI")
print("="*60)

# --- Asumsi De Moivre ---
omega_dm = 100
print(f"\n--- Asumsi De Moivre (ω={omega_dm}) ---")
print(f"Menghitung _{{{periode_hidup_non_bulat}}}p_{{{usia_awal}}}:")
hasil_dm_prob = mp.survival_prob_assumption(
    age=usia_awal,
    period=periode_hidup_non_bulat,
    interest_rate=suku_bunga, # Diperlukan untuk ActuarialCalculator
    assumption_type='de_moivre',
    params=[float(omega_dm)]
)
hasil_dm_prob.show()
print(f"   -> Verifikasi Manual: ({omega_dm} - {usia_awal} - {periode_hidup_non_bulat}) / ({omega_dm} - {usia_awal}) = {((omega_dm - usia_awal - periode_hidup_non_bulat) / (omega_dm - usia_awal)) if (omega_dm - usia_awal) != 0 else 0.0:.8f}")


# --- Asumsi Constant Force of Mortality (CFM) ---
mu_cfm = 0.025
print(f"\n--- Asumsi Constant Force of Mortality (μ={mu_cfm}) ---")
print(f"Menghitung _{{{periode_hidup_non_bulat}}}p_{{{usia_awal}}}:")
hasil_cfm_prob = mp.survival_prob_assumption(
    age=usia_awal,
    period=periode_hidup_non_bulat,
    interest_rate=suku_bunga,
    assumption_type='constant_mu_cfm',
    params=[mu_cfm]
)
hasil_cfm_prob.show()
print(f"   -> Verifikasi Manual: exp(-{mu_cfm} * {periode_hidup_non_bulat}) = {math.exp(-mu_cfm * periode_hidup_non_bulat):.8f}")


# --- Asumsi Gompertz ---
print(f"\n--- Asumsi Gompertz (B={B_gomp:.2e}, c={c_gomp}) ---")
print(f"Menghitung _{{{periode_hidup_non_bulat}}}p_{{{usia_awal}}}:")
hasil_gomp_prob = mp.survival_prob_assumption(
    age=usia_awal,
    period=periode_hidup_non_bulat,
    interest_rate=suku_bunga,
    assumption_type='gompertz',
    params=[B_gomp, c_gomp]
)
hasil_gomp_prob.show()
# Verifikasi manual untuk Gompertz _t p_x = exp( - B * c^x * (c^t - 1) / ln(c) )
if c_gomp == 1.0:
    expected_gomp_prob = math.exp(-B_gomp * periode_hidup_non_bulat)
else:
    integral_mu_t_gomp = B_gomp * (c_gomp**usia_awal) * (c_gomp**periode_hidup_non_bulat - 1) / math.log(c_gomp)
    expected_gomp_prob = math.exp(-integral_mu_t_gomp)
print(f"   -> Verifikasi Manual: {expected_gomp_prob:.8f}")


# --- Asumsi Makeham ---
print(f"\n--- Asumsi Makeham (A={A_make:.2e}, B={B_make:.2e}, c={c_make}) ---")
print(f"Menghitung _{{{periode_hidup_non_bulat}}}p_{{{usia_awal}}}:")
hasil_make_prob = mp.survival_prob_assumption(
    age=usia_awal,
    period=periode_hidup_non_bulat,
    interest_rate=suku_bunga,
    assumption_type='makeham',
    params=[A_make, B_make, c_make]
)
hasil_make_prob.show()
# Verifikasi manual untuk Makeham _t p_x = exp( - A*t - B * c^x * (c^t - 1) / ln(c) )
integral_A_t_make = A_make * periode_hidup_non_bulat
integral_Bc_part_t_make: float
if c_make == 1.0:
    integral_Bc_part_t_make = B_make * periode_hidup_non_bulat
else:
    integral_Bc_part_t_make = B_make * (c_make**usia_awal) * (c_make**periode_hidup_non_bulat - 1) / math.log(c_make)
expected_make_prob = math.exp(-(integral_A_t_make + integral_Bc_part_t_make))
print(f"   -> Verifikasi Manual: {expected_make_prob:.8f}")
```

```python
print("\n" + "="*60)
print("PENGUJIAN NSP WHOLE LIFE DENGAN BERBAGAI ASUMSI")
print("="*60)

# --- Asumsi Gompertz ---
print(f"\n--- NSP Whole Life (A_{{{usia_awal}}}) dengan Asumsi Gompertz (B={B_gomp:.2e}, c={c_gomp}) ---")
hasil_gomp_nsp = mp.nsp_wl_assumption(
    age=usia_awal,
    interest_rate=suku_bunga,
    assumption_type='gompertz',
    params=[B_gomp, c_gomp]
)
hasil_gomp_nsp.show()

# --- Asumsi Makeham ---
print(f"\n--- NSP Whole Life (A_{{{usia_awal}}}) dengan Asumsi Makeham (A={A_make:.2e}, B={B_make:.2e}, c={c_make}) ---")
hasil_make_nsp = mp.nsp_wl_assumption(
    age=usia_awal,
    interest_rate=suku_bunga,
    assumption_type='makeham',
    params=[A_make, B_make, c_make]
)
hasil_make_nsp.show()
```

```python
print("\n" + "="*60)
print("PENGUJIAN PV ANUITAS WHOLE LIFE DUE DENGAN BERBAGAI ASUMSI")
print("="*60)

# --- Asumsi Gompertz ---
print(f"\n--- PV Anuitas (ä_{{{usia_awal}}}) dengan Asumsi Gompertz (B={B_gomp:.2e}, c={c_gomp}) ---")
hasil_gomp_ann = mp.pv_annuity_due_wl_assumption(
    age=usia_awal,
    interest_rate=suku_bunga,
    assumption_type='gompertz',
    params=[B_gomp, c_gomp]
)
hasil_gomp_ann.show()

# --- Asumsi Makeham ---
print(f"\n--- PV Anuitas (ä_{{{usia_awal}}}) dengan Asumsi Makeham (A={A_make:.2e}, B={B_make:.2e}, c={c_make}) ---")
hasil_make_ann = mp.pv_annuity_due_wl_assumption(
    age=usia_awal,
    interest_rate=suku_bunga,
    assumption_type='makeham',
    params=[A_make, B_make, c_make]
)
hasil_make_ann.show()

print("\n" + "="*60)
print("SELESAI CONTOH PENGGUNAAN ASUMSI LANJUTAN")
print("="*60)
```
