```python
# ---------------------------------------------------------------------------
# Sel Tunggal untuk Jupyter Notebook - examples/hitung_peluang.ipynb
# Fokus: Hanya ActuarialResult yang di-render kaya, teks lain via print()
# ---------------------------------------------------------------------------

import mortapy as mp
import os
from IPython.display import display # Hanya display yang dibutuhkan jika show() sudah ada

# --- Setup Awal ---
CUSTOM_TABLE_FILENAME = "tabel_kustom_sederhana.csv"
USE_CUSTOM_TABLE = False 
CUSTOM_TABLE_PATH = ""
if USE_CUSTOM_TABLE:
    try:
        current_dir = os.getcwd() 
        base_project_dir = os.path.abspath(os.path.join(current_dir, "..")) 
        CUSTOM_TABLE_PATH = os.path.join(base_project_dir, 'mortapy', 'tables', CUSTOM_TABLE_FILENAME)
    except NameError: 
        CUSTOM_TABLE_PATH = os.path.join('..', 'mortapy', 'tables', CUSTOM_TABLE_FILENAME)

# Parameter Umum
usia_awal = 30
periode = 2 
suku_bunga = 0.05
gender_pilihan = 'pria'

print("="*70)
print("MENGUJI SEMUA OPSI PENGGUNAAN MORTAPY (UNTUK USIA BULAT)")
print("="*70)
print(f"Parameter Umum: Usia = {usia_awal}, Periode = {periode} tahun, Gender = {gender_pilihan}, Suku Bunga = {suku_bunga*100:.2f}%")
print("-"*70)

# --- Opsi 1: Menggunakan Tabel Mortalita Default ---
print("\n--- OPSI 1: Menggunakan Tabel Mortalita Default (Indonesia) ---")
try:
    print(f"Menghitung {periode}p{usia_awal} untuk {gender_pilihan} menggunakan tabel default:")
    hasil_default_prob = mp.calculate_survival_prob_integer(
        age=usia_awal,
        period_years=periode,
        interest_rate=suku_bunga,
        gender=gender_pilihan
    )
    hasil_default_prob.show() # Panggil .show() untuk render LaTeX
except FileNotFoundError:
    print(f"ERROR: Tabel default tidak ditemukan. Pastikan path di api.py sudah benar.")
except Exception as e:
    print(f"Terjadi error pada Opsi 1: {e}")

# --- Opsi 2: Menggunakan Tabel Mortalita Kustom (Dinonaktifkan) ---
if USE_CUSTOM_TABLE:
    print("\n\n--- OPSI 2: Menggunakan Tabel Mortalita Kustom ---")
    try:
        if os.path.exists(CUSTOM_TABLE_PATH):
            tabel_kustom = mp.MortalityTable(CUSTOM_TABLE_PATH)
            print(f"Menghitung {periode}p{usia_awal} untuk {gender_pilihan} menggunakan tabel kustom ('{CUSTOM_TABLE_FILENAME}'):")
            hasil_kustom_prob = mp.calculate_survival_prob_integer(
                age=usia_awal,
                period_years=periode,
                interest_rate=suku_bunga,
                gender=gender_pilihan,
                mortality_table=tabel_kustom
            )
            hasil_kustom_prob.show() # Panggil .show()
        else:
            print(f"Peringatan: File tabel kustom '{CUSTOM_TABLE_FILENAME}' tidak ditemukan di '{CUSTOM_TABLE_PATH}'. Lewati tes ini.")
    except Exception as e:
        print(f"Terjadi error saat menggunakan tabel kustom (Opsi 2): {e}")
else:
    print(f"\n\n--- OPSI 2: Menggunakan Tabel Mortalita Kustom (Dinonaktifkan) ---")

# --- Opsi 3: Menggunakan base_qx ---
print("\n\n--- OPSI 3: Menggunakan `base_qx` (Asumsi Distribusi Teoretis) ---")
qx_teoretis = 0.01
try:
    print(f"Menghitung {periode}p{usia_awal} dengan asumsi q_x = {qx_teoretis}:")
    hasil_base_qx_prob = mp.calculate_survival_prob_integer(
        age=usia_awal,
        period_years=periode,
        interest_rate=suku_bunga,
        base_qx=qx_teoretis
    )
    hasil_base_qx_prob.show() # Panggil .show()
    print(f"   -> Verifikasi Manual: (1 - {qx_teoretis})^{periode} = {(1 - qx_teoretis)**periode:.8f}")
except Exception as e:
    print(f"Terjadi error pada Opsi 3: {e}")

# --- Opsi 4: Menggunakan base_px ---
print("\n\n--- OPSI 4: Menggunakan `base_px` (Asumsi Distribusi Teoretis) ---")
px_teoretis = 0.985
try:
    print(f"Menghitung {periode}p{usia_awal} dengan asumsi p_x = {px_teoretis}:")
    hasil_base_px_prob = mp.calculate_survival_prob_integer(
        age=usia_awal,
        period_years=periode,
        interest_rate=suku_bunga,
        base_px=px_teoretis
    )
    hasil_base_px_prob.show() # Panggil .show()
    print(f"   -> Verifikasi Manual: ({px_teoretis})^{periode} = {px_teoretis**periode:.8f}")
except Exception as e:
    print(f"Terjadi error pada Opsi 4: {e}")

print("\n\n--- PENGUJIAN FUNGSI API LAINNYA ---")
usia_lain = 40
try:
    print(f"\nMenghitung NSP Whole Life (A_{usia_lain}) untuk {gender_pilihan} dengan tabel default:")
    nsp_default_pria = mp.calculate_whole_life_nsp(age=usia_lain, interest_rate=suku_bunga, gender=gender_pilihan)
    nsp_default_pria.show() # Panggil .show()

    print(f"\nMenghitung PV Anuitas Whole Life Due (ä_{usia_lain}) untuk wanita dengan tabel default:")
    ann_default_wanita = mp.calculate_whole_life_annuity_pv(age=usia_lain, interest_rate=suku_bunga, gender='wanita')
    ann_default_wanita.show() # Panggil .show()

    print(f"\nMenghitung NSP Whole Life (A_{usia_lain}) dengan base_qx=0.02:")
    nsp_base_qx = mp.calculate_whole_life_nsp(age=usia_lain, interest_rate=suku_bunga, base_qx=0.02)
    nsp_base_qx.show() # Panggil .show()
except Exception as e:
    print(f"Terjadi error pada pengujian fungsi API lainnya: {e}")

print("\n" + "="*70)
print("SELESAI PENGUJIAN SEMUA OPSI")
print("="*70)
```
