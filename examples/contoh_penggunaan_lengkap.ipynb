```python
import mortapy as mp
import os
# from IPython.display import Markdown, display # display mungkin masih berguna untuk .show()

# Path ke tabel kustom (opsional)
# CUSTOM_TABLE_FILENAME = "tabel_kustom_sederhana.csv"
# CUSTOM_TABLE_PATH = os.path.join(os.path.dirname(__file__), '..', 'mortapy', 'tables', CUSTOM_TABLE_FILENAME)
USE_CUSTOM_TABLE = False 
CUSTOM_TABLE_PATH = ""
if USE_CUSTOM_TABLE:
    try:
        current_dir = os.getcwd()
        base_project_dir = os.path.abspath(os.path.join(current_dir, "..")) 
        CUSTOM_TABLE_PATH = os.path.join(base_project_dir, 'mortapy', 'tables', CUSTOM_TABLE_FILENAME)
    except NameError: 
        CUSTOM_TABLE_PATH = os.path.join('..', 'mortapy', 'tables', CUSTOM_TABLE_FILENAME)

# Parameter Umum
usia_awal = 30
periode = 2 # Untuk _2p_30
suku_bunga = 0.05
gender_pilihan = 'pria'

print("="*70)
print("MENGUJI SEMUA OPSI PENGGUNAAN MORTAPY (UNTUK USIA BULAT)")
print("="*70)
print(f"Parameter Umum: Usia = {usia_awal}, Periode = {periode} tahun, Gender = {gender_pilihan}, Suku Bunga = {suku_bunga*100:.2f}%")
print("-"*70)
```

```python
print("\n--- OPSI 1: Menggunakan Tabel Mortalita Default (Indonesia) ---")
try:
    print(f"Menghitung {periode}p{usia_awal} untuk {gender_pilihan} menggunakan tabel default:")
    hasil_default_prob = mp.survival_prob_table(
        age=usia_awal, 
        n_years=periode, 
        interest_rate=suku_bunga,
        gender=gender_pilihan
    )
    hasil_default_prob.show() # atau cukup 'hasil_default_prob' jika ini satu-satunya output yang diinginkan dari sel ini
except FileNotFoundError:
    print(f"ERROR: Tabel default tidak ditemukan. Pastikan path di api_tables.py sudah benar.")
except Exception as e:
    print(f"Terjadi error pada Opsi 1: {e}")
```

```python
if USE_CUSTOM_TABLE:
    print("\n\n--- OPSI 2: Menggunakan Tabel Mortalita Kustom ---")
    try:
        if os.path.exists(CUSTOM_TABLE_PATH):
            tabel_kustom = mp.MortalityTable(CUSTOM_TABLE_PATH) # Pastikan mp.MortalityTable diekspor di __init__.py
            print(f"Menghitung {periode}p{usia_awal} untuk {gender_pilihan} menggunakan tabel kustom ('{CUSTOM_TABLE_FILENAME}'):")
            hasil_kustom_prob = mp.survival_prob_table(
                age=usia_awal,
                n_years=periode,
                interest_rate=suku_bunga,
                gender=gender_pilihan,
                mortality_table=tabel_kustom
            )
            hasil_kustom_prob.show()
        else:
            print(f"Peringatan: File tabel kustom '{CUSTOM_TABLE_FILENAME}' tidak ditemukan di '{CUSTOM_TABLE_PATH}'. Lewati tes ini.")
    except Exception as e:
        print(f"Terjadi error saat menggunakan tabel kustom (Opsi 2): {e}")
else:
    print(f"\n\n--- OPSI 2: Menggunakan Tabel Mortalita Kustom (Dinonaktifkan) ---")
```

```python
print("\n\n--- OPSI 3: Menggunakan `base_qx` (Asumsi Distribusi Teoretis) ---")
qx_teoretis = 0.01
try:
    print(f"Menghitung {periode}p{usia_awal} dengan asumsi q_x = {qx_teoretis}:")
    hasil_base_qx_prob = mp.survival_prob_assumption(
        age=usia_awal,
        period=float(periode),
        interest_rate=suku_bunga,
        assumption_type='constant_qx',
        param1=qx_teoretis
    )
    hasil_base_qx_prob.show()
    print(f"   -> Verifikasi Manual: (1 - {qx_teoretis})^{periode} = {(1 - qx_teoretis)**periode:.8f}")
except Exception as e:
    print(f"Terjadi error pada Opsi 3: {e}")
```

```python
print("\n\n--- OPSI 4: Menggunakan `base_px` (Asumsi Distribusi Teoretis) ---")
px_teoretis = 0.985
try:
    print(f"Menghitung {periode}p{usia_awal} dengan asumsi p_x = {px_teoretis}:")
    hasil_base_px_prob = mp.survival_prob_assumption(
        age=usia_awal,
        period=float(periode),
        interest_rate=suku_bunga,
        assumption_type='constant_px',
        param1=px_teoretis
    )
    hasil_base_px_prob.show()
    print(f"   -> Verifikasi Manual: ({px_teoretis})^{periode} = {px_teoretis**periode:.8f}")
except Exception as e:
    print(f"Terjadi error pada Opsi 4: {e}")
```

```python
# Lanjutan dari sel-sel sebelumnya di Jupyter Notebook Anda
# Pastikan 'import mortapy as mp' sudah dijalankan.
# Jika ActuarialResult dan metode .show() belum mengimpor IPython.display,
# Anda mungkin perlu 'from IPython.display import display, Math' di sini.

print("\n" + "="*50)
print("CONTOH PERHITUNGAN 2p30 DENGAN ASUMSI DE MOIVRE & CFM")
print("="*50)

# Parameter dasar untuk contoh ini
usia_awal_contoh = 30
periode_contoh = 2.0 # Gunakan float untuk konsistensi dengan tipe parameter 'period'
suku_bunga_contoh = 0.05 # Diperlukan untuk ActuarialCalculator, meski tidak dipakai di p_x

# --- 1. Menggunakan Asumsi Hukum De Moivre ---
omega_untuk_de_moivre = 110 # Tentukan usia maksimum untuk De Moivre
print(f"\n--- Menghitung {int(periode_contoh)}p{usia_awal_contoh} dengan Asumsi De Moivre (ω = {omega_untuk_de_moivre}) ---")

try:
    prob_de_moivre = mp.survival_prob_assumption(
        age=usia_awal_contoh,
        period=periode_contoh,
        interest_rate=suku_bunga_contoh,
        assumption_type='de_moivre',
        param1=float(omega_untuk_de_moivre) # param1 adalah omega
    )
    prob_de_moivre.show() # Ini akan merender LaTeX: {}_{2}p_{30} (ω=110) = HASIL
    
    # Verifikasi Manual
    expected_value_dm = (omega_untuk_de_moivre - usia_awal_contoh - periode_contoh) / (omega_untuk_de_moivre - usia_awal_contoh)
    print(f"   -> Verifikasi Manual: ({omega_untuk_de_moivre} - {usia_awal_contoh} - {periode_contoh}) / ({omega_untuk_de_moivre} - {usia_awal_contoh}) = {expected_value_dm:.8f}")

except Exception as e:
    print(f"Terjadi error saat perhitungan De Moivre: {e}")


# --- 2. Menggunakan Asumsi Constant Force of Mortality (CFM) ---
mu_konstan_untuk_cfm = 0.015 # Tentukan nilai mu (force of mortality konstan)
print(f"\n\n--- Menghitung {int(periode_contoh)}p{usia_awal_contoh} dengan Asumsi CFM (μ = {mu_konstan_untuk_cfm}) ---")

try:
    prob_cfm = mp.survival_prob_assumption(
        age=usia_awal_contoh,
        period=periode_contoh,
        interest_rate=suku_bunga_contoh,
        assumption_type='constant_mu_cfm',
        param1=mu_konstan_untuk_cfm # param1 adalah mu
    )
    prob_cfm.show() # Ini akan merender LaTeX: {}_{2}p_{30} (μ=0.015) = HASIL

    # Verifikasi Manual
    import math # Pastikan math sudah diimpor di awal notebook/skrip Anda
    expected_value_cfm = math.exp(-mu_konstan_untuk_cfm * periode_contoh)
    print(f"   -> Verifikasi Manual: exp(-{mu_konstan_untuk_cfm} * {periode_contoh}) = {expected_value_cfm:.8f}")

except Exception as e:
    print(f"Terjadi error saat perhitungan CFM: {e}")

print("\n" + "="*50)
print("SELESAI CONTOH TAMBAHAN")
print("="*50)
```

```python
print("\n\n--- PENGUJIAN FUNGSI API LAINNYA ---")
usia_lain = 40
try:
    print(f"\nMenghitung NSP Whole Life (A_{usia_lain}) untuk {gender_pilihan} dengan tabel default:")
    nsp_default_pria = mp.nsp_wl_table(age=usia_lain, interest_rate=suku_bunga, gender=gender_pilihan)
    nsp_default_pria.show()

    print(f"\nMenghitung PV Anuitas Whole Life Due (ä_{usia_lain}) untuk wanita dengan tabel default:")
    ann_default_wanita = mp.pv_annuity_due_wl_table(age=usia_lain, interest_rate=suku_bunga, gender='wanita')
    ann_default_wanita.show()

    print(f"\nMenghitung NSP Whole Life (A_{usia_lain}) dengan base_qx=0.02:")
    nsp_base_qx = mp.nsp_wl_assumption(age=usia_lain, interest_rate=suku_bunga, assumption_type='constant_qx', param1=0.02)
    nsp_base_qx.show()
except Exception as e:
    print(f"Terjadi error pada pengujian fungsi API lainnya: {e}")

print("\n" + "="*70)
print("SELESAI PENGUJIAN SEMUA OPSI")
print("="*70)
```

```python
# Lanjutan dari sel-sel sebelumnya di Jupyter Notebook Anda
# Pastikan 'import mortapy as mp' sudah dijalankan.
# Jika ActuarialResult dan metode .show() belum mengimpor IPython.display,
# Anda mungkin perlu 'from IPython.display import display, Math' di sini.

print("\n" + "="*50)
print("CONTOH OPERASI ARITMATIKA PADA HASIL ACTUARIALRESULT")
print("="*50)

# --- Parameter untuk contoh operasi ---
usia_operasi_1 = 45
usia_operasi_2 = 50
suku_bunga_operasi = 0.04
gender_operasi = 'pria'
qx_asumsi_ops = 0.015

# --- Membuat beberapa objek ActuarialResult ---

print(f"\n1. Menghitung NSP Whole Life (A_{{{usia_operasi_1}}}) untuk {gender_operasi} dengan tabel default:")
nsp_tabel_1 = mp.nsp_wl_table(
    age=usia_operasi_1,
    interest_rate=suku_bunga_operasi,
    gender=gender_operasi
)
nsp_tabel_1.show() # Ini akan merender LaTeX

print(f"\n2. Menghitung NSP Whole Life (A_{{{usia_operasi_2}}}) dengan asumsi q_x = {qx_asumsi_ops}:")
nsp_asumsi_1 = mp.nsp_wl_assumption(
    age=usia_operasi_2,
    interest_rate=suku_bunga_operasi,
    assumption_type='constant_qx',
    param1=qx_asumsi_ops
)
nsp_asumsi_1.show() # Ini akan merender LaTeX

print(f"\n3. Menghitung PV Anuitas Whole Life Due (ä_{{{usia_operasi_1}}}) untuk {gender_operasi} dengan tabel default:")
pv_annuity_tabel_1 = mp.pv_annuity_due_wl_table(
    age=usia_operasi_1,
    interest_rate=suku_bunga_operasi,
    gender=gender_operasi
)
pv_annuity_tabel_1.show() # Ini akan merender LaTeX

print("\n" + "-"*50)
print("MELAKUKAN OPERASI ARITMATIKA")
print("-" * 50)

# --- Penjumlahan dua ActuarialResult ---
print("\nOperasi: NSP Tabel 1 + NSP Asumsi 1")
hasil_penjumlahan = nsp_tabel_1 + nsp_asumsi_1
hasil_penjumlahan.show()

# --- Pengurangan ActuarialResult dengan angka ---
angka_pengurang = 0.05
print(f"\nOperasi: PV Anuitas Tabel 1 - {angka_pengurang}")
hasil_pengurangan_angka = pv_annuity_tabel_1 - angka_pengurang
hasil_pengurangan_angka.show()

# --- Penjumlahan angka dengan ActuarialResult (Reflektif) ---
angka_penambah = 0.1
print(f"\nOperasi: {angka_penambah} + NSP Tabel 1")
hasil_jumlah_reflektif = angka_penambah + nsp_tabel_1
hasil_jumlah_reflektif.show()

# --- Kombinasi Lebih Kompleks ---
print("\nOperasi Kombinasi: (NSP Tabel 1 + 0.02) - NSP Asumsi 1")
hasil_kompleks = (nsp_tabel_1 + 0.02) - nsp_asumsi_1
hasil_kompleks.show()

print("\n" + "="*50)
print("SELESAI CONTOH OPERASI ARITMATIKA")
print("="*50)
```
